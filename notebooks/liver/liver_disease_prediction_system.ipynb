```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import pickle

# 1. Load data and drop duplicates
df_clean = df.drop_duplicates().copy()

# 2. Target mapping: Original target 'Dataset' has 1 (Liver disease) and 2 (No disease)
# Standard ML standard is 1 (Positive/Disease) and 0 (Negative/No disease)
df_clean['Dataset'] = df_clean['Dataset'].map({1: 1, 2: 0})

# Split Features and Target
X = df_clean.drop(columns=['Dataset'])
y = df_clean['Dataset']

# 3. Stratified Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Data points: Train={X_train.shape[0]}, Test={X_test.shape[0]}")

# 4. Feature columns identification
num_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()

print("Numerical columns:", num_cols)
print("Categorical columns:", cat_cols)

# 5. Preprocessing Pipelines
num_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

cat_pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', drop='first'))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_pipeline, num_cols),
    ('cat', cat_pipeline, cat_cols)
])

# 6. Define Candidate Models
models = {
    'Logistic Regression': LogisticRegression(random_state=42, max_iter=1000),
    'Random Forest': RandomForestClassifier(random_state=42),
    'Gradient Boosting': GradientBoostingClassifier(random_state=42)
}

# Evaluate Models with Stratified Cross-Validation
cv_results = {}
scoring = ['accuracy', 'f1', 'roc_auc']
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for name, clf in models.items():
    # Assemble preprocessing and estimator in a single Pipeline
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', clf)
    ])

    scores = cross_validate(pipeline, X_train, y_train, cv=cv, scoring=scoring)
    cv_results[name] = {
        'Mean CV Accuracy': scores['test_accuracy'].mean(),
        'Mean CV F1': scores['test_f1'].mean(),
        'Mean CV ROC-AUC': scores['test_roc_auc'].mean()
    }

cv_df = pd.DataFrame(cv_results).T
display(cv_df)
```

```python
# 7. Train and evaluate the final best model (e.g. Gradient Boosting or Logistic Regression)
best_model_name = cv_df['Mean CV ROC-AUC'].idxmax()
print(f"Best model selected based on CV ROC-AUC: {best_model_name}")

best_classifier = models[best_model_name]

# Complete End-to-End Pipeline
final_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', best_classifier)
])

# Fit on entire training data
final_pipeline.fit(X_train, y_train)

# Evaluate on Test set
y_pred = final_pipeline.predict(X_test)
y_proba = final_pipeline.predict_proba(X_test)[:, 1]

print(f"\n--- Test Set Metrics for {best_model_name} ---")
print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test, y_proba):.4f}")

# 8. Export the pipeline object to a pickle file
model_output_path = 'liver_disease_pipeline.pkl'
with open(model_output_path, 'wb') as f:
    pickle.dump(final_pipeline, f)
print(f"\nSaved final fitted model pipeline at: '{model_output_path}' ✅")
```
